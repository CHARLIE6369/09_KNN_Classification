```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split,KFold,cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

## 2. Data collection

```python
wine_data = load_wine()
wine_data.data,wine_data.feature_names,wine_data.target
wine_data_df = pd.DataFrame(wine_data.data,columns=wine_data.feature_names)
wine_data_df["target"] = wine_data.target
```

## 4. Data Preparation

```python
x = wine_data_df.drop("target",axis=1)
y = wine_data_df["target"]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
stand_scal =StandardScaler()
```

```python
std_x= stand_scal.fit_transform(x)
```

## 5. Model Building

```python
X_train, X_test, y_train, y_test = train_test_split(std_x,y,test_size=0.20,random_state=12,stratify=y)
```

```python
knn_model = KNeighborsClassifier(n_neighbors=9)
```

## 6. Model Training

```python
knn_model.fit(X_train,y_train)
```

## 7. Model Testing

```python
y_train_pred = knn_model.predict(X_train)
```

```python
y_test_pred = knn_model.predict(X_test)
```

## 8. Model Evaluation

#### When K = 3

```python
knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train,y_train)
y_train_pred = knn_model.predict(X_train)
print("Accuracy_score:",round(accuracy_score(y_train,y_train_pred),4))
```

#### When K = 5

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train,y_train)
y_train_pred = knn_model.predict(X_train)
print("Accuracy_score:",round(accuracy_score(y_train,y_train_pred),4))
```

#### When K = 7

```python
knn_model = KNeighborsClassifier(n_neighbors=7)
knn_model.fit(X_train,y_train)
y_train_pred = knn_model.predict(X_train)
print("Accuracy_score:",round(accuracy_score(y_train,y_train_pred),4))
```

#### When K = 9

```python
knn_model = KNeighborsClassifier(n_neighbors=9)
knn_model.fit(X_train,y_train)
y_train_pred = knn_model.predict(X_train)
print("Accuracy_score:",round(accuracy_score(y_train,y_train_pred),4))
```

## kNeighbors fing Technique

```python
kflod = KFold(n_splits=5,shuffle=True, random_state=12)
cross_score = []
for i in range (3,52,2):
    knn_classifier = KNeighborsClassifier(n_neighbors=i)
    knn_classifier.fit(X_train,y_train)
    cross_scores = cross_val_score(estimator = knn_classifier,X=std_x,y=y,cv=kflod)
    cross_score.append(cross_scores.mean())
```
